```python
df_jobs = pd.read_csv('./data/jobs.csv')
```

```python
df_jobs.columns
```

```python
# count how many null entries there are in each field
for c in df_jobs.columns:
    print(c, len(df_jobs[df_jobs[c].isna()]))
```

```python
# simple hot deck imputation: the field "is_current" can be inferred from other fields
has_startend = (df_jobs.is_current.isna()) & ~(df_jobs.started_on.isna()) & ~(df_jobs.ended_on.isna())
df_jobs.loc[has_startend,'is_current'] = df_jobs.loc[has_startend,'is_current'].fillna('f')
```
