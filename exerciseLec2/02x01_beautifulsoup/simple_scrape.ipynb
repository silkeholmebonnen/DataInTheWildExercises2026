## Imports

```python
import requests # http requests
import re # regular expressions
from bs4 import BeautifulSoup # xml parsing
```

## HTTP request

```python
# send a HTTP GET request and store the response
url = 'https://en.wikipedia.org/wiki/James_Hetfield'
response = requests.get(url)
status_code = response.status_code
if status_code != 200:
    raise Exception(f'Error {status_code} ({requests.status_codes._codes[status_code]})')
else:
    html_text = response.text
    print(f'Request successful')
```

## Parse html content

### With regular expressions

```python
# find artist name
# compile the regular expression for fast matching
re_artistname = re.compile('<td class="infobox-data nickname">(.+?)<\/td>')
# find all matches
matches = re.findall(re_artistname, html_text)
print('Name:')
print(f'found {len(matches)} match')
for m in matches:
    print(m)

# EXERCISE: extract the list of genres using a regular expression
# HINT: use a first regular expression to extract the section of interest
# then, apply a second regular expression to the result to extract the list of genres.
# Remember that adding a ? on a quantifier (?, * or +) makes the regular expression non-greedy (lazy),
# i.e., it will stop a the first match it finds.
#
# solution is at the bottom of the notebook
print('\nGenres:')
# YOUR CODE HERE
print(matches)
```

### With BeautifulSoup

```python
# parse the response with BeautifulSoup
soup = BeautifulSoup(html_text)
```

```python
# find artist name
name_section = soup.find('td', {'class':'infobox-data nickname'})
print(name_section.text)

# find associated genres
genre_section = soup.find_all('td', {'class':'infobox-data'})[2]
genres = genre_section.find_all('a')
print([g['title'] for g in genres])

# EXERCISE: extract the list of bands James Hetfield was a member of before Metallica
# YOUR CODE HERE
```
