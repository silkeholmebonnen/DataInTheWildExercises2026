## CSS and XPatch practice

```python
import requests
from scrapy.selector import Selector
```

```python
# download a page of interest
url = 'https://…'
response = requests.get(url)
html_text = response.text
```

```python
# or use this simplified example
html_text = """
<html>
  <head><title>Welcome</title></head>
  <body>
    <h1>My page</h1>
    <div class="contact">
      check out my
      <a href="www.mypage1.com">
        webpage1
      </a>
      <a href="www.mypage2.com">
        webpage2
      </a>
    </div>
  </body>
</html>
"""
```

```python
# create a Scrapy selector out of the html text
s = Selector(text=html_text)
```

```python
#CSS examples
print(s.css('body > div > a').getall())
print(s.css('body > h1::text').getall())
print(s.css('body > div::attr(class)').getall())
print(s.css('* title').getall())
print(s.css('* > div[class*="contact"]').getall())
print(s.css('* > .contact').getall())
print(s.css('body').css('h1').getall())
```

```python
print(s.xpath('/html/body/div/a').getall())
print(s.xpath('/html/body/h1/text()').getall())
print(s.xpath('/html/body/div/@class').getall())
print(s.xpath('.//title').getall())
print(s.xpath('/html/*/div[@class="contact"]').getall())
print(s.xpath('/html').xpath('//body').getall())
```
